```python
import pandas as pd
df = pd.read_csv('cleaned_I_THINK.csv')

df.describe()
```

```python
df.isnull().sum()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.hist(df['user_rating'], bins=30, edgecolor='black', color='blue')
plt.title('Analysis of Ratings')
plt.xlabel('Rating')
plt.ylabel('Count')
plt.show()
```

```python
import ast
df['genres'] = df['genres'].apply(ast.literal_eval)
all_genres = [genre for sublist in df['genres'] for genre in sublist]

genre_counts = pd.Series(all_genres).value_counts()

# Plot the genre distribution
plt.figure(figsize=(12, 8))
genre_counts.plot(kind='bar', color='blue')
plt.title('Genre Analysis')
plt.xlabel('Genres')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
# Plot number of releases per year
release_year_counts = df['release_year'].value_counts().sort_index()

plt.figure(figsize=(10, 6))
release_year_counts.plot(kind='line', color='blue')
plt.title('Movie Releases')
plt.xlabel('Year')
plt.ylabel('Number of Movies Released')
plt.show()
```

```python
director_counts = df['director'].value_counts().head(10)

plt.figure(figsize=(10, 6))
director_counts.plot(kind='barh', color='blue')
plt.title('Top Directors by Movies')
plt.xlabel('Number')
plt.ylabel('Director')
plt.show()
```

```python
# Create a df for genre and user ratings
genre_ratings = df.explode('genres').groupby('genres')['user_rating'].mean()

# Sort the average ratings by genre
genre_ratings_sorted = genre_ratings.sort_values(ascending=False)

plt.figure(figsize=(12, 8))
genre_ratings_sorted.plot(kind='bar', color='blue')
plt.title('Average User Rating by Genre Analysis')
plt.xlabel('Genres')
plt.ylabel('Average Rating')
plt.xticks(rotation=45, ha='right')
plt.show()
```

Insights Summary:

1) Ratings Analysis:
Most ratings are concentrated around specific values, showing overall viewer preferences.

2) Genre Distribution:
Popular genres like Action, Drama, or Comedy dominate, while others are less common.

3) Movie Releases Over Time:
Movie production has generally increased, with some peaks and drops over the years.

4) Top Directors:
The top 10 directors have made the most movies, showing their influence in the industry.

5) Average Ratings by Genre:
Certain genres (example History, War) have higher average ratings, indicating audience preference.
