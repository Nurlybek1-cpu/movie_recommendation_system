Preprocessing Steps Documentation:

1. Importing Required Libraries
pandas: Used for data manipulation and analysis.
MinMaxScaler from sklearn.preprocessing: Used for normalizing the numerical column user_rating.

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
```

2. Reading the Dataset:

The raw data is read into a pandas DataFrame named data from a CSV file (movies_data.csv).

```python
data = pd.read_csv('movies_data.csv')
```

3. Filtering Inappropriate Content:

A list of inappropriate words related to adult or sensitive content is defined.
A custom function filter_keywords_or_description() is created to check for inappropriate words in the keywords and description columns.
If any inappropriate word is found, the function replaces the content of that row with None.

```python

inappropriate_words = [
    "lgbtq", "prostitute", "erotic", "nudity", "nude", "porn", "xxx", "explicit", 
    "sensual", "intimate", "seduction", "fetish", "seductive", "temptation", "flirtatious", "voyeurism", "depravity", "lecherous", "forbidden",  "sensational", "obscenities", "bare", "exposed body", "indulgent" "kinky", "provocative", "raunchy", "sultry", "taboo",  "sex", "sexual", "voyeur", "bdsm", "stripper",  "uncensored", "provocation", "escort", "affair", "orgy", "brothel", "lust", "vulgar", "softcore", "hardcore", "molestation", "incest", "rape", "pedophile", "perversion", "depraved", "lewd", 
    "pornographic", "adult movie", "taboo themes", "indecent", "exposed",
   
]

# filter adult
def filter_keywords_or_description(text):
    if isinstance(text, str):
        for word in inappropriate_words:
            if word in text.lower():
                return None  # Return None or empty string to filter out
    return text

# Apply filter to columns
data['keywords'] = data['keywords'].apply(filter_keywords_or_description)
data['description'] = data['description'].apply(filter_keywords_or_description)
```

4. Handling Missing Values

For keywords and description: Missing or filtered values (None) are replaced with the string 'Unknown'.
For genres and director: Any missing values are replaced with 'Unknown'.

```python
data['keywords'] = data['keywords'].fillna('Unknown')
data['description'] = data['description'].fillna('Unknown')
data['genres'] = data['genres'].fillna('Unknown')
data['director'] = data['director'].fillna('Unknown')
```

5. One-Hot Encoding for Genres

The genres column is a string with multiple genres separated by commas.
Steps:

Split the genres string into a list using .str.split(', ').
Use the explode() function to flatten the list into rows.

Apply one-hot encoding using str.get_dummies().
Recombine the encoded genres back into the original DataFrame using groupby(level=0).sum().

```python
# One-hot encoding 
data['genres'] = data['genres'].str.split(', ')
genres_one_hot = data['genres'].explode().str.get_dummies().groupby(level=0).sum()
data = pd.concat([data, genres_one_hot], axis=1)
```

6. Normalizing User Ratings:

The user_rating column is normalized using MinMaxScaler to scale values between 0 and 1.

```python
# Normalizing 
scaler = MinMaxScaler()
data['user_rating'] = scaler.fit_transform(data[['user_rating']])
```

7. Handling Missing Values for Critical Columns:

Rows with NaN values in the title or user_rating columns are dropped, as these fields are critical.

```python
# Dropping null rows for title and user_rating
data = data.dropna(subset=['title', 'user_rating'])
```

8. Removing Duplicate Titles:

```python
# Drop duplicate title
data = data.drop_duplicates(subset='title', keep='first')
```

9. Saving the Cleaned Dataset:

```python
output = "cleaned_I_THINK.csv"
data.to_csv(output, index=False)
print(f"Cleaned file saved as: {output}")

model = pd.read_csv(output)
model.head()
```
